```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Loading, Inspecting, and Displaying two Images
uploaded = files.upload()
image_names = list(uploaded.keys())

img1 = cv2.imread(image_names[0])
img2 = cv2.imread(image_names[1])

img1 = cv2.resize(img1, (400, 400))
img2 = cv2.resize(img2, (400, 400))

print("Image 1 Shape:", img1.shape)
print("Image 2 Shape:", img2.shape)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
axes[0].set_title('Image 1 Original')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
axes[1].set_title('Image 2 Original')
axes[1].axis('off')
plt.show()


# Step 2: Manipulate image matrices directly at the array level
# Step 3: Crop 30x30 from first image random location and paste in on second image. (Name this image as Image3)
h2, w2, _ = img2.shape
rand_y = np.random.randint(0, h2 - 30)
rand_x = np.random.randint(0, w2 - 30)

crop_from_img2 = img2[rand_y:rand_y+30, rand_x:rand_x+30]

img3 = img1.copy()
center_y = (img1.shape[0] - 30) // 2
center_x = (img1.shape[1] - 30) // 2
img3[center_y:center_y+30, center_x:center_x+30] = crop_from_img2

plt.figure(figsize=(5, 5))
plt.imshow(cv2.cvtColor(img3, cv2.COLOR_BGR2RGB))
plt.title('Image3')
plt.axis('off')
plt.show()


# Step 4: Convert the two images to gray, also Image3
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
gray3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(gray1, cmap='gray')
axes[0].set_title('Gray Image 1')
axes[0].axis('off')

axes[1].imshow(gray2, cmap='gray')
axes[1].set_title('Gray Image 2')
axes[1].axis('off')

axes[2].imshow(gray3, cmap='gray')
axes[2].set_title('Gray Image 3')
axes[2].axis('off')
plt.show()


# Step 5: Apply kernel that sharp the edge of Image3
sharpen_kernel = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]])

sharpened_Image3 = cv2.filter2D(gray3, -1, sharpen_kernel)

plt.figure(figsize=(5, 5))
plt.imshow(sharpened_Image3, cmap='gray')
plt.title('Sharpened Image3')
plt.axis('off')
plt.show()
```
